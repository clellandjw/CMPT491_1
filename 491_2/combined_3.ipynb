```python
bad_names = players[
    players["name"].str.contains("Ã|Â|â", regex=True, na=False)
]

print("Rows with suspicious names:", len(bad_names))
print(bad_names["name"].drop_duplicates().head(20).tolist())
```

```python
for name in bad_names["name"].drop_duplicates().head(20):
    print(name)
```

```python
def fix_name(name):
    try:
        return name.encode("latin1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return name
```

```python
for name in bad_names["name"].drop_duplicates().head(20):
    print(fix_name(name))
```

```python
players_clean = players.copy()

players_clean["name"] = players_clean["name"].apply(fix_name)
```

```python
remaining = players_clean[
    players_clean["name"].str.contains("Ã|Â|â", regex=True, na=False)
]

print("Remaining suspicious rows:", len(remaining))
```

```python
fixture_sides = players_clean[
    ["season", "fixture", "was_home", "opponent_team"]
].drop_duplicates()

print(fixture_sides.shape)
print(fixture_sides.head(10))
```

```python
team_lookup = fixture_sides.merge(
    fixture_sides,
    on=["season", "fixture"],
    suffixes=("", "_other")
)

team_lookup = team_lookup[
    team_lookup["was_home"] != team_lookup["was_home_other"]
]

print(team_lookup.shape)
print(team_lookup.head(10))
```

```python
team_lookup = team_lookup[
    ["season", "fixture", "was_home", "opponent_team_other"]
].rename(columns={
    "opponent_team_other": "correct_team_id"
})

print(team_lookup.head(10))
print(team_lookup.shape)
```

```python
players_corrected = players_clean.merge(
    team_lookup,
    on=["season", "fixture", "was_home"],
    how="left",
    validate="many_to_one"
)

print(players_corrected.shape)
print("Missing team IDs:", players_corrected["correct_team_id"].isna().sum())
```

```python
historical = players_corrected[
    players_corrected["season"].isin([
        "2016-17", "2017-18", "2018-19", "2019-20"
    ])
].copy()

historical["team_numeric"] = pd.to_numeric(
    historical["team"], errors="coerce"
)

mismatches = historical[
    historical["team_numeric"] != historical["correct_team_id"]
]

print("Historical appearances:", len(historical))
print("Mismatched team IDs:", len(mismatches))

print(
    mismatches[
        ["season", "fixture", "name", "team",
         "team_name", "correct_team_id"]
    ].head(15).to_string(index=False)
)
```

```python
print(
    mismatches.groupby("season").size()
)

print(
    "\nUnique players affected:",
    mismatches.groupby(
        ["season", "element"]
    ).ngroups
)
```
