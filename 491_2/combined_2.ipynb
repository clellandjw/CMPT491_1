```python
all_seasons = []

for season, files in season_files.items():
    gw_file, player_file = files

    # Load the two files for this season
    gw = pd.read_csv(
        f"raw_player_data/{gw_file}",
        encoding="latin1"
    )

    player_info = pd.read_csv(
        f"raw_player_data/{player_file}",
        encoding="latin1"
    )

    # Older files do not have the player's team,
    # so get it from the player information file
    if "team" not in gw.columns:
        season_merged = gw.merge(
            player_info[["id", "team"]],
            left_on="element",
            right_on="id",
            how="left"
        )
    else:
        # Newer files already contain team
        season_merged = gw.copy()

    # Identify which season these rows came from
    season_merged["season"] = season

    # Store this completed season
    all_seasons.append(season_merged)

    # Check our work
    print(
        season,
        season_merged.shape,
        season_merged["team"].isna().sum()
    )
```
