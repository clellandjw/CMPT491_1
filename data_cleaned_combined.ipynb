```python
common_columns = None

for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")
    
    if common_columns is None:
        common_columns = set(df.columns)
    else:
        common_columns = common_columns.intersection(df.columns)

print("Number of common columns:", len(common_columns))
print(sorted(common_columns))
```

```python
for i, column in enumerate(sorted(common_columns), start=1):
    print(i, column)
```

```python
df_2015 = pd.read_csv("2015.csv")

for year in range(2016, 2026):
    df = pd.read_csv(f"{year}.csv")
    
    missing = set(df_2015.columns) - set(df.columns)
    
    print(f"\n{year} - 2015 columns not found:")
    print(sorted(missing))
```

```python
comparison = []

columns_2015 = set(df_2015.columns)

for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")
    columns_year = set(df.columns)

    comparison.append({
        "Year": year,
        "Total Columns": len(columns_year),
        "Same as 2015": len(columns_2015 & columns_year),
        "Missing from 2015": len(columns_2015 - columns_year),
        "New vs 2015": len(columns_year - columns_2015)
    })

comparison_df = pd.DataFrame(comparison)

comparison_df
```

```python
required_columns = [
    "Date",
    "HomeTeam", "AwayTeam",
    "FTHG", "FTAG", "FTR",
    "HTHG", "HTAG", "HTR",
    "HS", "AS",
    "HST", "AST",
    "HC", "AC",
    "HF", "AF",
    "HY", "AY",
    "HR", "AR"
]
for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")

    missing = set(required_columns) - set(df.columns)

    print(year, "Missing:", missing)
```

```python
quality_check = []

for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")
    selected = df[required_columns]

    quality_check.append({
        "Year": year,
        "Rows": len(selected),
        "Missing Values": selected.isnull().sum().sum(),
        "Duplicate Rows": selected.duplicated().sum()
    })

quality_df = pd.DataFrame(quality_check)

quality_df
```

```python
for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")
    
    print(
        year,
        "FTR:", sorted(df["FTR"].unique()),
        "HTR:", sorted(df["HTR"].unique())
    )
    
```

```python
numeric_columns = [
    "FTHG", "FTAG",
    "HTHG", "HTAG",
    "HS", "AS",
    "HST", "AST",
    "HC", "AC",
    "HF", "AF",
    "HY", "AY",
    "HR", "AR"
]

for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")
    
    non_numeric = df[numeric_columns].select_dtypes(exclude="number").columns
    
    print(year, "Non-numeric columns:", list(non_numeric))
```

```python
all_numeric = []

for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")
    all_numeric.append(df[numeric_columns])

numeric_combined = pd.concat(all_numeric, ignore_index=True)

ranges = pd.DataFrame({
    "Minimum": numeric_combined.min(),
    "Maximum": numeric_combined.max()
})

ranges
```

Data correction: The Newcastle vs. West Ham match on 15/08/2021 contained an inconsistent value of 8 away shots and 9 away shots on target. Independent match records report 18 West Ham shots and 9 shots on target. The away-shots value was therefore corrected from 8 to 18.

```python
df_2021.loc[
    (df_2021["Date"] == "15/08/2021") &
    (df_2021["HomeTeam"] == "Newcastle") &
    (df_2021["AwayTeam"] == "West Ham"),
    "AS"
] = 18
```

```python
teams = set()

for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")

    teams.update(df["HomeTeam"].unique())
    teams.update(df["AwayTeam"].unique())

print("Number of unique team names:", len(teams))

for team in sorted(teams):
    print(team)
```

```python
for year in range(2015, 2026):
    df = pd.read_csv(f"{year}.csv")

    home_spaces = (df["HomeTeam"] != df["HomeTeam"].str.strip()).sum()
    away_spaces = (df["AwayTeam"] != df["AwayTeam"].str.strip()).sum()

    print(year, "Home spaces:", home_spaces, "Away spaces:", away_spaces)
```

```python
for year in range(2016, 2026):
    df = pd.read_csv(f"{year}.csv")

    parsed_dates = pd.to_datetime(
        df["Date"],
        dayfirst=True,
        errors="coerce"
    )

    print(
        year,
        "Invalid dates:", parsed_dates.isna().sum(),
        "Chronological:", parsed_dates.is_monotonic_increasing
    )
```

```python
for year in range(2016, 2026):
    season = f"{year}/{str(year + 1)[-2:]}"
    print(year, "->", season)
```

```python
all_seasons = []

for year in range(2016, 2026):
    df = pd.read_csv(f"{year}.csv")

    # Keep only the columns needed for our research questions
    df = df[required_columns].copy()

    # Create the season label
    df["Season"] = f"{year}/{str(year + 1)[-2:]}"

    all_seasons.append(df)

combined = pd.concat(all_seasons, ignore_index=True)

combined.shape
```

```python
combined["Date"] = pd.to_datetime(
    combined["Date"],
    dayfirst=True
)
```

```python
final_check = pd.DataFrame({
    "Rows": [len(combined)],
    "Columns": [len(combined.columns)],
    "Missing Values": [combined.isnull().sum().sum()],
    "Duplicate Rows": [combined.duplicated().sum()],
    "HST > HS": [(combined["HST"] > combined["HS"]).sum()],
    "AST > AS": [(combined["AST"] > combined["AS"]).sum()],
    "HTHG > FTHG": [(combined["HTHG"] > combined["FTHG"]).sum()],
    "HTAG > FTAG": [(combined["HTAG"] > combined["FTAG"]).sum()]
})

final_check
```
